**args-kwargs**

- args means we can provide multiple values without argument name
   - which means while calling function we can provide directly 'Cristiano' , 40
   - while define the function you need to mention argument name also
   - the output will comes in **tuple** format: ('Cristiano',40)
- kwargs means keyword arguments means we can provide multiple values with argument name
    - while define the finction no need to provide the name of the arguments
    - while call the functions definitely we need to provide with argument names
       - name = 'Cristiano' , age = 40
    - the out put comes in dict format: {'name': 'Cristiano', 'age' : 40}

```python
def fun1(*args):
    print(args)
fun1(10,20) 
# with out name
# output in tuple format
```

```python
def fun2(**kwargs):
    print(kwargs)
fun2(a=10,b=20)
# with name 
# output in dict format
```
