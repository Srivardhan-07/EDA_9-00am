**if-elif-else**

- when there is one condition is there
  - we can expect two answers 
     - in that first under **if** block
     - second answer under **else** block
- when there are two conditions
     - we can expect three answers
         - in that first answer under **if** block based on condition-1
         - second answer under **elif** block based condition-2
         - thied answer under **else** block

```python
if <conditioon>:
    <write your code>
elif<condition>:
    <write your code>
else:
    <write your code>
```

```python
user = eval(input('enter a number :'))
if user == 1 :
    print('one')
elif user == 2:
    print('two')
elif user == 3:
    print('three')
else:
    print('enter valid number')
```
